### 7. Fill null values with the previous value of the same column

```python
df = df.ffill()
df
```

### 8. Fill the remaining null values with 0

After the forward fill no nulls are left, so this acts as a safety net for any value the previous step could not carry forward (e.g. a null in the very first row).

```python
df = df.fillna(0)
df
```
