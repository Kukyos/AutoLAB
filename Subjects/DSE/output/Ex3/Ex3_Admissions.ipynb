# Ex 3 — Data Visualization

**Dataset:** University Admissions (admissions.csv)  
**Reg No:** URK24CS1021

```python
print("URK24CS1021")

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (7, 4.5)

df = pd.read_csv("admissions.csv")
df
```

```python
print("URK24CS1021")

df.dtypes
```

### 1. Scatter plot — does a higher GPA come with a higher SAT score?

```python
print("URK24CS1021")

sns.scatterplot(data=df, x="GPA", y="SAT_Score", hue="Admitted", style="Admitted", s=120)
plt.title("SAT Score vs GPA")
plt.show()
```

### 2. Box plot — how do ACT scores differ between admitted and rejected?

```python
print("URK24CS1021")

sns.boxplot(data=df, x="Admitted", y="ACT_Score", hue="Admitted", palette="Set2", legend=False)
plt.title("ACT Score by admission status")
plt.show()
```

### 3. Bar plot — does a stronger recommendation mean a better chance?

```python
print("URK24CS1021")

rate = (df.groupby("Recommendation_Strength")["Admitted"]
          .apply(lambda s: (s == "Yes").mean())
          .reindex(["Weak", "Moderate", "Strong"]))
rate
```

```python
print("URK24CS1021")

sns.barplot(x=rate.index, y=rate.values, hue=rate.index, palette="Blues_d", legend=False)
plt.ylabel("Admission rate")
plt.ylim(0, 1)
plt.title("Admission rate by recommendation strength")
plt.show()
```

### 4. Histogram — how many activities does a typical applicant have?

```python
print("URK24CS1021")

plt.hist(df["Extracurriculars"], bins=range(0, 7), edgecolor="black", align="left")
plt.xticks(range(0, 6))
plt.yticks(range(0, 6))
plt.xlabel("Number of extracurriculars")
plt.ylabel("Number of students")
plt.show()
```

### 5. Violin plot — the box plot, plus the shape of the distribution

```python
print("URK24CS1021")

sns.violinplot(data=df, x="Admitted", y="GPA", hue="Admitted", palette="Set3", legend=False)
plt.title("GPA distribution by admission status")
plt.show()
```
